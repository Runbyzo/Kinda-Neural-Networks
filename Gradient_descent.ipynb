```python
import numpy as np
```

```python
# input data
toes = np.array([8.5, 9.5, 9.9, 9.0])
wlrec = np.array([0.65, 0.8, 0.8, 0.9])
nfuns = np.array([1.2, 1.3, 0.5, 1.0])
inputs = np.array([toes[0], wlrec[0], nfuns[0]])

# network weights
ip_wght = np.array([[0.1, 0.2, -0.1], [-.1, .1, .9], [.1, .4, .1]])
hp_wght = np.array([[0.3, 1.1, -0.3], [.1, .2, 0], [0, 1.3, 0.1]])
two_layer_weights = [ip_wght, hp_wght]

# true value for prediction
true_val = [1, 0.245, -0.423]

# 2l neural network
def neural_network(inputs, weights):
    hid_lyr = inputs.dot(weights[0])
    pred = hid_lyr.dot(weights[1])
    return pred

test_pred = neural_network(inputs, two_layer_weights)

print(f'prediction: {test_pred}')
print(f'goal:       {true_val}')
```

```python
# mean squaered error
def MSE(predictions, true_values):
    assert(len(predictions) == len(true_values))
    n = len(predictions)
    result = 0
    for i in range(n):
        result += np.power(predictions[i] - true_values[i], 2)
    return result/n
```

```python
print(MSE(test_pred, true_val))
```
