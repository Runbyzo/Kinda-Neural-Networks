```python
weights = np.array([0.1, 0.2, 0])
def neural_network(inputs, weights):
    return inputs.dot(weights)
```

```python
def elem_mult(num, vector):
    output = [0, 0, 0]
    assert(len(vector) == len(output))

    for i in range(len(vector)):
        output[i] = num * vector[i]

    return output

def neural_network_1(inputs, weights):
    return elem_mult(inputs, weights)
```

```python
def mat_mul(vector, matrix):
    assert(len(vector) == len(matrix))
    output = [0, 0, 0]
    for i in range(len(vector)):
        output[i] = vector.dot(matrix[i])
    return output

def neural_network_2(inputs, weights):
    return mat_mul(inputs, weights)
```

3 in 3 out

```python
print(neural_network_2(inputs, weights))
```
